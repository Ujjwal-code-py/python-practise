```python
## fibbonaci
n=int(input("Enter the number of terms: "))
a=0
b=1
fib=[]
if(n==0):
    fib.append(a)
elif(n==1):  
    fib.append(a)
    fib.append(b)
else:
    fib.append(a)
    fib.append(b)  
    for i in range(2,n):
        c=a+b
    
        fib.append(c)
        a=b
        b=c
    
print(fib)
```

```python
arr = [10, 20, 4, 45, 99]
first=float('-inf')
secound=float('inf')
for num in arr:
    if(num>first):
        secound=first
        first=num
    elif( (num>secound) and (num!=first)):
        secound=num
print(secound)
        
```

```python
s1="silent"
s2="lisent"

if(len(s1)!=len(s2)):
    print("Not anagram")
    
else:
    freq={}
    is_anagram=True

    for ch in s1:
        freq[ch]=freq.get(ch,0)+1
        
    for ch in s2:
        freq[ch]=freq.get(ch,0)-1
    
    for value in freq.values():
        if(value != 0):
            is_anagram=False
            break
    if(is_anagram):
        print("Anagram")
    else:
        print("not anagram")
    
```

```python
s="madam"
l=0
h=len(s)-1
is_palindrome=True
while(l<h):
    if(s[l]!=s[h]):
        is_palindrome=False
        break
    l+=1
    h-=1
if(is_palindrome):
    print("Palindrome")     
    
else:
    print("Not Palindrome")
```
