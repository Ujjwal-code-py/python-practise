#### Find first and last occurence in sorted Array

```python
arr=[1,2,3,4,4,4,5,6,6,7,8,9,9,10]
target=int(input())
first=-1
last=-1
n=len(arr)

for i in range(n):
    
    if(arr[i]==target):
        if(first==-1):
            first=i
    if(arr[i]>target):
        last=i-1
        break
print([first,last])
        
    
            
        
    ## O(n) time complexity
    ## O(1) space complexity
```

```python
## optimal solution
class Solution:

    def lowerBound(self, nums, target):
        low, high = 0, len(nums) - 1
        lb = -1

        while low <= high:
            mid = (low + high) // 2

            if nums[mid] >= target:
                if nums[mid] == target:
                    lb = mid
                high = mid - 1
            else:
                low = mid + 1

        return lb

    def upperBound(self, nums, target):
        low, high = 0, len(nums) - 1
        ub = -1

        while low <= high:
            mid = (low + high) // 2

            if nums[mid] > target:
                ub = mid
                high = mid - 1
            else:
                low = mid + 1

        return ub

    def searchRange(self, nums, target):

        lb = self.lowerBound(nums, target)

        if lb == -1:
            return [-1, -1]

        ub = self.upperBound(nums, target)

        if ub == -1:
            return [lb, len(nums) - 1]

        return [lb, ub - 1]


# Example
obj = Solution()

nums = [5, 7, 7, 8, 8, 10]
target = 8

print(obj.searchRange(nums, target))

## Tc=o(2logn)
```
