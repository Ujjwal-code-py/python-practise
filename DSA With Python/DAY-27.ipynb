#### Find first and last occurence in sorted Array

```python
arr=[1,2,3,4,4,4,5,6,6,7,8,9,9,10]
target=int(input())
first=-1
last=-1
n=len(arr)

for i in range(n):
    
    if(arr[i]==target):
        if(first==-1):
            first=i
    if(arr[i]>target):
        last=i-1
        break
print([first,last])
        
    
            
        
    ## O(n) time complexity
    ## O(1) space complexity
```

```python
class solution:
    def __init__(self):
        pass
    def lower_bound(arr,target):
        n=len(arr)
        low=0
        high=n-1
        lb=-1
        while low<=high:
            mid=(low+high)//2
            if(arr[mid]>=target):
                lb=mid
                high=mid-1
            else:
                low=mid+1
        return lb

    def upper_bound(arr,target):
        n=len(arr)
        low=0
        high=n-1
        lb=-1
        while low<=high:
            mid=(low+high)//2
            if(arr[mid]>target):
                ub=mid
                high=mid-1
            else:
                low=mid+1
        return ub
    def ans(self,arr,target):
        lb=self.lower_bound(arr,target)
        ub=self.upper_bound(arr,target)
        return[lb,ub-1]
```
