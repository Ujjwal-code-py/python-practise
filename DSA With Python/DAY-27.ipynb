#### Find first and last occurence in sorted Array

```python
arr=[1,2,3,4,4,4,5,6,6,7,8,9,9,10]
target=int(input())
first=-1
last=-1
n=len(arr)

for i in range(n):
    
    if(arr[i]==target):
        if(first==-1):
            first=i
    if(arr[i]>target):
        last=i-1
        break
print([first,last])
        
    
            
        
    
```
