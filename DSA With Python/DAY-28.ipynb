Count the Occurances of number

```python
## brute force
nums=[1,2,3,4,4,4,4,4,4,4,5,6,7,7,8,8]
target=3
n=len(nums)
count=0
for i in range(n):
    if(nums[i]==target):
        count+=1
print(count)
    
```

```python
def lower_bound(nums,target):
    n=len(nums)
    low=0
    high=n-1
    lb=-1
    while(low<=high):
        mid=(low+high)//2
        if(nums[mid]>=target):
            if nums[mid] == target:
                lb=mid
            high=mid-1
        else:
            low=mid+1
    return lb
            
def upper_bound(nums,target):
    n=len(nums)
    low=0
    high=n-1
    ub=n
    while(low<=high):
        mid=(low+high)//2
        if(nums[mid]>target):
            ub=mid
            high=mid-1
        else:
            low=mid+1
    return ub  
```

```python
nums=[1,2,3,4,4,4,4,4,4,4,5,6,7,7,8,8]
target=4
lb=lower_bound(nums=nums,target=target)
ub=upper_bound(nums=nums,target=target)
print(ub-lb)
```
