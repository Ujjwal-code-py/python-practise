### Search insert position

```python
## Array is sorted
## return the correct index of target value should be
arr=[1,3,4,5,6,7,9,11,12,23,34]
target=int(input())
n=len(arr)

for i in range(n):
    if(target<=arr[i]):
        print(i)
        break
        
## TC=o(n)
```

```python
## optimal
arr=[1,3,4,5,6,7,9,11,12,23,34]
target=int(input())
n=len(arr)
low=0
high=(n-1)
lb=n
while(low<=high):
    mid=(low+high)//2
    if(arr[mid]>=target):
        lb=mid
        high=mid-1
    else:
        low=mid+1
print(lb)

```
